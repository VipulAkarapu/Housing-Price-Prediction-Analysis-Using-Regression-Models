# Importing required libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

# Importing the dataset

```python
dataset = pd.read_csv('/content/kc_house_data.csv')
X = dataset.iloc[:, 3:4].values
y = dataset.iloc[:, 2:3].values
```

```python
print(X)
```

```python
print(y)
```

# Reducing the data

```python
required_test_size = 0.03
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = required_test_size, random_state = 0)
```

```python
X = X_test
y = y_test
```

```python
#y = y.reshape(len(y),1)
```

```python
print(y)
```

# Feature Scaling

```python
from sklearn.preprocessing import StandardScaler
sc_X = StandardScaler()
sc_y = StandardScaler()
X = sc_X.fit_transform(X)
y = sc_y.fit_transform(y)
```

```python
print(X)
```

```python
print(y)
```

# Training the SVR model on the whole dataset

```python
from sklearn.svm import SVR
regressor = SVR(kernel = 'rbf')
regressor.fit(X, y)
```

# Predicting a new result for the flat of 6 bedrooms

```python
sc_y.inverse_transform(regressor.predict(sc_X.transform([[6]])).reshape(-1,1))
```

# Visualising the SVR results

```python
plt.scatter(sc_X.inverse_transform(X), sc_y.inverse_transform(y), color = 'yellow')
plt.plot(sc_X.inverse_transform(X), sc_y.inverse_transform(regressor.predict(X).reshape(-1,1)), color = 'green')
plt.title('Truth or Bluff (SVR)')
plt.xlabel('no of bedrooms')
plt.ylabel('Price ')
plt.show()
```
